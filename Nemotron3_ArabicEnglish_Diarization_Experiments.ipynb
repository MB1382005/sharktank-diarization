```python
import torch
from transformers import AutoModelForAudioFrameClassification, AutoProcessor

device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.bfloat16 if (device == "cuda" and torch.cuda.is_bf16_supported()) else torch.float16

MODEL_ID = "nvidia/Nemotron-3-Diarization"
processor = AutoProcessor.from_pretrained(MODEL_ID)
model = AutoModelForAudioFrameClassification.from_pretrained(MODEL_ID, torch_dtype=dtype).to(device)
model.eval()

TARGET_SR = processor.feature_extractor.sampling_rate
print(f"Model loaded on {device} ({dtype}) | Sampling Rate = {TARGET_SR} Hz")
```

```python
import soundfile as sf
import numpy as np
from pyannote.core import Annotation, Segment
from pyannote.metrics.diarization import DiarizationErrorRate, JaccardErrorRate

def run_diarization(wav_path):
    audio, sr = sf.read(str(wav_path), dtype="float32")
    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    processor.set_streaming_mode("low_latency")

    def chunk_generator():
        yield processor(
            audio[:processor.num_samples_first_audio_chunk],
            sampling_rate=TARGET_SR,
            is_streaming=True,
            is_first_audio_chunk=True,
        )
        mel_idx = processor.num_mel_frames_per_step
        start_idx = processor.audio_chunk_start(mel_idx)
        while (end_idx := start_idx + processor.num_samples_per_audio_chunk) <= audio.shape[0]:
            yield processor(
                audio[start_idx:end_idx],
                sampling_rate=TARGET_SR,
                is_streaming=True,
                is_first_audio_chunk=False,
            )
            mel_idx += processor.num_mel_frames_per_step
            start_idx = processor.audio_chunk_start(mel_idx)

        if start_idx < audio.shape[0]:
            yield processor(
                audio[start_idx:],
                sampling_rate=TARGET_SR,
                is_streaming=True,
                is_first_audio_chunk=False,
                is_last_audio_chunk=True,
            )

    speaker_cache, all_logits = None, []
    with torch.inference_mode():
        for inputs in chunk_generator():
            inputs = inputs.to(model.device, dtype=model.dtype)
            outputs = model(**inputs, speaker_cache=speaker_cache)
            all_logits.append(outputs.logits.cpu())
            speaker_cache = outputs.speaker_cache

    full_logits = torch.cat(all_logits, dim=1)
    segments = processor.extract_speaker_dict(full_logits)[0]
    return segments, len(audio) / TARGET_SR


def load_ground_truth_rttm(rttm_path):
    ref = Annotation(uri=Path(rttm_path).stem)
    with open(rttm_path, "r", encoding="utf-8") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 8 and parts[0] == "SPEAKER":
                start, dur, spk = float(parts[3]), float(parts[4]), parts[7]
                if dur > 0:
                    ref[Segment(start, start + dur)] = spk
    return ref


def compute_metrics(rttm_path, predicted_segments):
    ref = load_ground_truth_rttm(rttm_path)
    hyp = Annotation(uri=Path(rttm_path).stem)
    for seg in predicted_segments:
        s, e = float(seg["Start"]), float(seg["End"])
        if e > s:
            hyp[Segment(s, e)] = f"speaker_{seg['Speaker']}"

    der_metric = DiarizationErrorRate(collar=0.0, skip_overlap=False)
    der_25_metric = DiarizationErrorRate(collar=0.25, skip_overlap=False)
    jer_metric = JaccardErrorRate(collar=0.0, skip_overlap=False)

    comp = der_metric(ref, hyp, detailed=True)
    comp25 = der_25_metric(ref, hyp, detailed=True)
    jer_val = jer_metric(ref, hyp)
    total = max(comp["total"], 1e-9)

    return {
        "ref_speakers": len(ref.labels()),
        "pred_speakers": len(hyp.labels()),
        "DER_0.0s_%": round(comp["diarization error rate"] * 100, 2),
        "Missed_%": round((comp["missed detection"] / total) * 100, 2),
        "FalseAlarm_%": round((comp["false alarm"] / total) * 100, 2),
        "Confusion_%": round((comp["confusion"] / total) * 100, 2),
        "DER_0.25s_%": round(comp25["diarization error rate"] * 100, 2),
        "JER_%": round(jer_val * 100, 2),
    }

print("Helper functions ready!")
```

```python
import pandas as pd
from scipy.signal import butter, sosfiltfilt
from scipy.ndimage import uniform_filter1d

# 1. دالة الـ Audio Preprocessing (التجربة 1)
def preprocess_arabic_english_audio(audio: np.ndarray, sr: int = 16000) -> np.ndarray:
    x = audio - np.mean(audio)
    sos = butter(4, [80.0, 7600.0], btype="bandpass", fs=sr, output="sos")
    x = sosfiltfilt(sos, x).astype(np.float32)

    win = int(0.05 * sr)  # 50 ms
    env = np.sqrt(uniform_filter1d(x ** 2, size=win) + 1e-10)
    noise_floor = np.percentile(env, 20)
    target_rms = np.percentile(env, 85)

    speech_mask = np.clip((env - 2.0 * noise_floor) / (noise_floor + 1e-8), 0.0, 1.0)
    desired_gain = np.clip(target_rms / (env + 1e-6), 1.0, 3.0)
    gain = 1.0 + speech_mask * (desired_gain - 1.0)

    x_boosted = x * gain
    peak = np.max(np.abs(x_boosted)) + 1e-8
    x_norm = x_boosted / peak
    x_comp = np.tanh(1.4 * x_norm) / np.tanh(1.4)
    return (0.95 * x_comp).astype(np.float32)


def run_diarization_custom(wav_path, use_preproc=False):
    audio, sr = sf.read(str(wav_path), dtype="float32")
    if audio.ndim > 1:
        audio = audio.mean(axis=1)

    if use_preproc:
        audio = preprocess_arabic_english_audio(audio, sr=TARGET_SR)

    processor.set_streaming_mode("low_latency")

    def chunk_generator():
        yield processor(
            audio[:processor.num_samples_first_audio_chunk],
            sampling_rate=TARGET_SR,
            is_streaming=True,
            is_first_audio_chunk=True,
        )
        mel_idx = processor.num_mel_frames_per_step
        start_idx = processor.audio_chunk_start(mel_idx)
        while (end_idx := start_idx + processor.num_samples_per_audio_chunk) <= audio.shape[0]:
            yield processor(
                audio[start_idx:end_idx],
                sampling_rate=TARGET_SR,
                is_streaming=True,
                is_first_audio_chunk=False,
            )
            mel_idx += processor.num_mel_frames_per_step
            start_idx = processor.audio_chunk_start(mel_idx)

        if start_idx < audio.shape[0]:
            yield processor(
                audio[start_idx:],
                sampling_rate=TARGET_SR,
                is_streaming=True,
                is_first_audio_chunk=False,
                is_last_audio_chunk=True,
            )

    speaker_cache, all_logits = None, []
    with torch.inference_mode():
        for inputs in chunk_generator():
            inputs = inputs.to(model.device, dtype=model.dtype)
            outputs = model(**inputs, speaker_cache=speaker_cache)
            all_logits.append(outputs.logits.cpu())
            speaker_cache = outputs.speaker_cache

    full_logits = torch.cat(all_logits, dim=1)
    segments = processor.extract_speaker_dict(full_logits)[0]
    return segments, full_logits


# 2. التشغيل على أول عينتين بس (2 Samples) ومقارنة Exp 0 vs Exp 1
TWO_SAMPLES = test_manifest[:2]
results_2samples = []
cached_logits = {}  # هنحفظ الـ logits هنا عشان نجرب عليها التجربة 2 في ثانية واحدة من غير إعادة تشغيل الموديل!

for item in TWO_SAMPLES:
    rid = item["recording_id"]
    w_path = DATA_DIR / item["audio_path"]
    r_path = DATA_DIR / item["rttm_path"]

    print(f"--- Processing {rid} ---")
    # Exp 0: Baseline
    segs_base, logits_base = run_diarization_custom(w_path, use_preproc=False)
    m_base = compute_metrics(r_path, segs_base)
    results_2samples.append({"Sample": rid, "Experiment": "Exp 0: Baseline", **m_base})

    # Exp 1: Audio Preprocessing
    segs_pre, logits_pre = run_diarization_custom(w_path, use_preproc=True)
    m_pre = compute_metrics(r_path, segs_pre)
    results_2samples.append({"Sample": rid, "Experiment": "Exp 1: + Preprocessing", **m_pre})

    cached_logits[rid] = {"base": logits_base, "pre": logits_pre, "rttm": r_path}

df_2samples = pd.DataFrame(results_2samples)
print("\n=== Per-Sample Results (2 Samples) ===")
display(df_2samples)

print("\n=== Average across 2 Samples (Exp 0 vs Exp 1) ===")
display(df_2samples.groupby("Experiment").mean(numeric_only=True).round(2))
```

```python
# تطبيق الـ 3 معادلات الرياضية (Math Post-Processing) على الـ cached_logits المحفوظة
def apply_diarization_math(
    logits: torch.Tensor,
    theta_on: float = 0.40,          # عتبة بداية الكلام
    theta_off: float = 0.18,         # عتبة استمرار الكلام (Hysteresis)
    pad_onset: float = 0.10,         # توسيع بداية الجملة (ثانية)
    pad_offset: float = 0.15,        # توسيع نهاية الجملة (ثانية)
    max_breath_gap: float = 0.85,    # سد فجوات النَفَس لنفس المتحدث (Morphological Closing)
    min_turn_dur: float = 0.15,      # حذف النبضات القصيرة جداً
    min_spk_energy_sec: float = 15.0 # حذف/دمج المتحدث الرابع الوهمي (Phantom Projection)
) -> list[dict]:
    probs = torch.sigmoid(logits[0].float()).cpu().numpy()  # (T, 8), step = 10ms
    num_frames, num_spks = probs.shape
    dt = 0.01  # 10 ms

    raw_by_spk = {}
    total_dur_by_spk = {}

    for spk in range(num_spks):
        p = probs[:, spk]
        # 1. Hysteresis Dual-Thresholding (theta_on / theta_off)
        active = np.zeros(num_frames, dtype=bool)
        is_on = False
        for t in range(num_frames):
            if not is_on and p[t] >= theta_on:
                is_on = True
            elif is_on and p[t] < theta_off:
                is_on = False
            active[t] = is_on

        if not np.any(active):
            continue

        padded = np.concatenate([[False], active, [False]])
        starts = np.where(padded[1:] & ~padded[:-1])[0] * dt
        ends = np.where(~padded[1:] & padded[:-1])[0] * dt

        # Boundary Dilation
        segs = []
        for s, e in zip(starts, ends):
            if (e - s) < 0.05:
                continue
            s_pad = max(0.0, float(s - pad_onset))
            e_pad = min(num_frames * dt, float(e + pad_offset))
            segs.append([s_pad, e_pad])

        if not segs:
            continue

        # 2. 1D Temporal Morphological Closing (سد فجوات النَفَس لنفس المتحدث)
        merged = [segs[0]]
        for s, e in segs[1:]:
            if s - merged[-1][1] <= max_breath_gap:
                merged[-1][1] = max(merged[-1][1], e)
            else:
                merged.append([s, e])

        merged = [[s, e] for s, e in merged if (e - s) >= min_turn_dur]
        if merged:
            raw_by_spk[spk] = merged
            total_dur_by_spk[spk] = sum(e - s for s, e in merged)

    # 3. Phantom-Channel Energy Projection (دمج المتحدث الرابع الوهمي)
    major_spks = [k for k, d in total_dur_by_spk.items() if d >= min_spk_energy_sec]
    if not major_spks and total_dur_by_spk:
        major_spks = list(total_dur_by_spk.keys())

    final_segments = []
    for spk, segs in raw_by_spk.items():
        if spk in major_spks:
            for s, e in segs:
                final_segments.append({"Start": s, "End": e, "Speaker": spk})
        else:
            for s, e in segs:
                f1 = max(0, int(s / dt))
                f2 = min(num_frames, int(e / dt))
                best_major = max(major_spks, key=lambda m: float(probs[f1:f2, m].mean()))
                final_segments.append({"Start": s, "End": e, "Speaker": best_major})

    final_segments.sort(key=lambda x: (x["Start"], x["End"]))
    return final_segments


# --- مقارنة الـ 4 تجارب في ثانية واحدة ---
math_rows = list(results_2samples)  # فيها Exp 0 و Exp 1

for rid, data in cached_logits.items():
    r_path = data["rttm"]

    # Exp 2: الماث لوحده على الصوت الخام
    segs_math_only = apply_diarization_math(data["base"])
    m_math = compute_metrics(r_path, segs_math_only)
    math_rows.append({"Sample": rid, "Experiment": "Exp 2: + Math Only", **m_math})

    # Exp 3: Preprocessing + الماث مع بعض
    segs_pre_math = apply_diarization_math(data["pre"])
    m_pre_math = compute_metrics(r_path, segs_pre_math)
    math_rows.append({"Sample": rid, "Experiment": "Exp 3: Preproc + Math", **m_pre_math})

df_math = pd.DataFrame(math_rows)
order = ["Exp 0: Baseline", "Exp 1: + Preprocessing", "Exp 2: + Math Only", "Exp 3: Preproc + Math"]
df_math["Experiment"] = pd.Categorical(df_math["Experiment"], categories=order, ordered=True)
df_math = df_math.sort_values(["Sample", "Experiment"])

print("=== Per-Sample Comparison (All 4 Experiments) ===")
display(df_math)

print("\n=== Average across 2 Samples (Exp 0 vs Exp 1 vs Exp 2 vs Exp 3) ===")
display(df_math.groupby("Experiment", observed=False).mean(numeric_only=True).round(2))
```

```python
import torch.nn as nn
import torch.nn.functional as F
from scipy.optimize import linear_sum_assignment

# 1. تجميد الـ 31 طبقة بتاعة Nemotron-3 بالكامل (Freeze Backbone)
for param in model.parameters():
    param.requires_grad = False

frozen_params = sum(p.numel() for p in model.parameters())
print(f"Frozen Nemotron-3 Backbone Parameters: {frozen_params:,} (100% Frozen)")


# 2. بناء طبقة الـ Fine-Tuning الجديدة (Multi-Scale Temporal Residual Adapter Head)
class TemporalResidualAdapter(nn.Module):
    """
    Lightweight trainable head on top of frozen Nemotron-3 10ms logits.
    Uses multi-scale dilated 1D convolutions (10ms, 50ms, 250ms, 750ms context)
    to learn Egyptian Arabic-English turn continuity and boundary refinement.
    """
    def __init__(self, num_speakers: int = 8, hidden: int = 24):
        super().__init__()
        # Per-speaker temporal context convs (shared across speaker channels so it generalizes to any speaker order!)
        self.conv_in = nn.Conv1d(1, hidden, kernel_size=5, padding=2)
        self.conv_d1 = nn.Conv1d(hidden, hidden, kernel_size=5, padding=10, dilation=5)    # ~200ms context
        self.conv_d2 = nn.Conv1d(hidden, hidden, kernel_size=5, padding=50, dilation=25)   # ~1.0s breath context
        self.conv_out = nn.Conv1d(hidden, 1, kernel_size=3, padding=1)

        # Cross-speaker interaction layer (helps suppress phantom 4th speaker)
        self.cross_spk = nn.Conv1d(num_speakers, num_speakers, kernel_size=1)

        # Initialize residual output near zero so training starts cleanly from the pretrained model!
        nn.init.zeros_(self.conv_out.weight)
        nn.init.zeros_(self.conv_out.bias)
        nn.init.eye_(self.cross_spk.weight.squeeze(-1))
        nn.init.zeros_(self.cross_spk.bias)
        self.alpha = nn.Parameter(torch.tensor(0.5))

    def forward(self, logits: torch.Tensor) -> torch.Tensor:
        # logits: (B, T, S=8)
        B, T, S = logits.shape
        x = logits.permute(0, 2, 1).reshape(B * S, 1, T)  # (B*S, 1, T)
        h = F.gelu(self.conv_in(x))
        h = F.gelu(self.conv_d1(h)) + h
        h = F.gelu(self.conv_d2(h)) + h
        delta_temporal = self.conv_out(h).reshape(B, S, T)  # (B, S, T)

        refined = self.cross_spk(logits.permute(0, 2, 1)) + self.alpha * delta_temporal
        return refined.permute(0, 2, 1)  # (B, T, S)


adapter = TemporalResidualAdapter().to(device)
trainable_params = sum(p.numel() for p in adapter.parameters() if p.requires_grad)
print(f"Trainable Adapter Head Parameters: {trainable_params:,}")


# 3. تحويل ملفات الـ RTTM لمصفوفة أهداف 10ms ومطابقتها بـ Permutation-Invariant Hungarian Matching (PIL)
def build_aligned_10ms_targets(logits_tensor: torch.Tensor, rttm_path: Path) -> torch.Tensor:
    probs = torch.sigmoid(logits_tensor[0].float()).cpu().numpy()  # (T, 8)
    T, S = probs.shape
    dt = 0.01

    ref = load_ground_truth_rttm(rttm_path)
    ref_labels = sorted(ref.labels())
    ref_matrix = np.zeros((T, len(ref_labels)), dtype=np.float32)

    for seg, _, spk in ref.itertracks(yield_label=True):
        idx = ref_labels.index(spk)
        f1 = max(0, min(T, int(round(seg.start / dt))))
        f2 = max(0, min(T, int(round(seg.end / dt))))
        ref_matrix[f1:f2, idx] = 1.0

    # Match each ground-truth speaker to the best model output channel (PIL alignment)
    cost_matrix = -np.matmul(probs.T, ref_matrix)  # (8, K)
    row_ind, col_ind = linear_sum_assignment(cost_matrix)

    aligned_target = np.zeros((T, S), dtype=np.float32)
    for r, c in zip(row_ind, col_ind):
        aligned_target[:, r] = ref_matrix[:, c]

    return torch.tensor(aligned_target, dtype=torch.float32).unsqueeze(0)  # (1, T, 8)


# تجهيز داتا تدريب الـ Adapter من ملفات الـ DEV
train_pairs = []
for rid, d in dev_cached.items():
    z = d["logits"].float().to(device)
    y = build_aligned_10ms_targets(d["logits"], d["rttm"]).to(device)
    train_pairs.append((rid, z, y))


# 4. حلقة التدريب (Fine-Tuning Loop بـ AdamW + PIL BCE + Dice Loss + Phantom Suppression)
optimizer = torch.optim.AdamW(adapter.parameters(), lr=3e-3, weight_decay=1e-4)

def pil_dice_bce_loss(pred_logits, targets):
    bce = F.binary_cross_entropy_with_logits(pred_logits, targets)
    probs = torch.sigmoid(pred_logits)
    inter = (probs * targets).sum(dim=1)
    dice = 1.0 - ((2.0 * inter + 1.0) / (probs.sum(dim=1) + targets.sum(dim=1) + 1.0)).mean()
    return 0.6 * bce + 0.4 * dice

print("\nFine-tuning Temporal Residual Adapter on ArzEn DEV...")
adapter.train()
for epoch in range(1, 41):
    total_loss = 0.0
    for rid, z, y in train_pairs:
        optimizer.zero_grad()
        out_logits = adapter(z)
        loss = pil_dice_bce_loss(out_logits, y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    if epoch % 10 == 0 or epoch == 1:
        print(f"  Epoch {epoch:02d}/40 | Loss: {total_loss / len(train_pairs):.4f}")

adapter.eval()

# 5. معايرة سريعة للـ Threshold على مخرجات الـ Adapter المعموله Fine-Tuning (على الـ DEV)
dev_ft_cached = {}
with torch.inference_mode():
    for rid, d in dev_cached.items():
        ft_logits = adapter(d["logits"].float().to(device)).cpu()
        dev_ft_cached[rid] = {"logits": ft_logits, "rttm": d["rttm"]}

best_ft_params = dict(best_params)
best_ft_der, _ = eval_params_on_set(dev_ft_cached, best_ft_params)
for th_on in [0.35, 0.40, 0.45, 0.50, 0.55]:
    for th_off in [0.18, 0.22, 0.25, 0.30, 0.35]:
        for gap in [0.75, 0.90, 1.05, 1.20]:
            p = dict(best_ft_params, theta_on=th_on, theta_off=th_off, max_breath_gap=gap, pad_onset=0.08, pad_offset=0.14)
            d_val, _ = eval_params_on_set(dev_ft_cached, p)
            if d_val < best_ft_der:
                best_ft_der = d_val
                best_ft_params = p

print(f"\nFine-Tuned Adapter Best DEV DER: {best_ft_der:.2f}% (was {best_dev_der:.2f}% before FT)")


# 6. تقييم الموديل بعد الـ Fine-Tuning على عينات الـ TEST (C03-R04 & C03-R08)
all_5_rows = [r for r in df_final.to_dict("records") if r["Experiment"] != "Exp 5: Frozen-Backbone Fine-Tuned"]

with torch.inference_mode():
    for rid, data in cached_logits.items():
        ft_test_logits = adapter(data["pre"].float().to(device)).cpu()
        segs_ft = apply_diarization_math(ft_test_logits, **best_ft_params)
        m_ft = compute_metrics(data["rttm"], segs_ft)
        all_5_rows.append({"Sample": rid, "Experiment": "Exp 5: Frozen-Backbone Fine-Tuned", **m_ft})

df_5 = pd.DataFrame(all_5_rows)
print("\n=== Complete Ablation Study on TEST Samples (Exp 0 -> Exp 5) ===")
display(df_5.groupby("Experiment", sort=False).mean(numeric_only=True).round(2))
```

```python
from scipy.ndimage import gaussian_filter1d

def compute_10ms_energy(wav_path: Path, num_frames: int, sr: int = 16000) -> np.ndarray:
    """Compute normalized 10ms acoustic RMS energy envelope aligned with model frames."""
    audio, _ = sf.read(str(wav_path), dtype="float32")
    if audio.ndim > 1:
        audio = audio.mean(axis=1)
    hop = int(0.01 * sr)  # 10 ms = 160 samples
    win = int(0.03 * sr)  # 30 ms window
    rms = np.sqrt(uniform_filter1d(audio ** 2, size=win) + 1e-10)
    frame_rms = rms[::hop]
    if len(frame_rms) < num_frames:
        frame_rms = np.pad(frame_rms, (0, num_frames - len(frame_rms)), mode="edge")
    else:
        frame_rms = frame_rms[:num_frames]

    p20, p85 = np.percentile(frame_rms, 20), np.percentile(frame_rms, 85)
    norm_e = np.clip((frame_rms - p20) / (p85 - p20 + 1e-8), 0.0, 1.0)
    return norm_e


def apply_energy_collision_free_math(
    logits: torch.Tensor,
    wav_path: Path,
    theta_on: float = 0.44,
    theta_off: float = 0.24,
    smooth_sigma: float = 4.0,       # تنعيم جاوسي (40ms) لمنع التقطيع اللحظي
    energy_boost: float = 0.08,      # دعم الاحتمال بطاقة الصوت الحقيقية
    pad_onset: float = 0.07,
    pad_offset: float = 0.13,
    max_breath_gap: float = 1.25,
    min_turn_dur: float = 0.15,
    min_spk_energy_sec: float = 15.0,
) -> list[dict]:
    probs = torch.sigmoid(logits[0].float()).cpu().numpy()  # (T, 8)
    T, S = probs.shape
    dt = 0.01

    # 1. Gaussian temporal smoothing on 10ms probabilities
    if smooth_sigma > 0:
        probs = gaussian_filter1d(probs, sigma=smooth_sigma, axis=0)

    # 2. Acoustic-Energy Conditioning: boost active speaker slightly when true acoustic energy is high
    energy = compute_10ms_energy(wav_path, T, TARGET_SR)  # (T,)
    top_spk_at_t = np.argmax(probs, axis=1)
    top_prob_at_t = np.max(probs, axis=1)

    # Only boost the dominant speaker at frame t if its prob is already lukewarm (> 0.12) and audio has real energy
    for t in range(T):
        dom = top_spk_at_t[t]
        if probs[t, dom] > 0.12:
            probs[t, dom] = min(1.0, probs[t, dom] + energy_boost * energy[t])

    # 3. Hysteresis Dual-Thresholding + Phantom Projection
    raw_by_spk = {}
    total_dur_by_spk = {}

    for spk in range(S):
        p = probs[:, spk]
        active = np.zeros(T, dtype=bool)
        is_on = False
        for t in range(T):
            if not is_on and p[t] >= theta_on:
                is_on = True
            elif is_on and p[t] < theta_off:
                is_on = False
            active[t] = is_on

        if not np.any(active):
            continue

        padded = np.concatenate([[False], active, [False]])
        starts = np.where(padded[1:] & ~padded[:-1])[0]
        ends = np.where(~padded[1:] & padded[:-1])[0]

        # 4. Collision-Free Padding & Acoustic Silence Trimming
        pad_on_f = int(round(pad_onset / dt))
        pad_off_f = int(round(pad_offset / dt))
        segs = []
        for s_f, e_f in zip(starts, ends):
            if (e_f - s_f) < 5:
                continue
            # Expand onset only while no OTHER speaker is strongly dominant (> 0.55) and energy > 0.05
            new_s = s_f
            for f in range(s_f - 1, max(0, s_f - pad_on_f) - 1, -1):
                other_max = np.max(np.delete(probs[f], spk))
                if other_max > 0.55:
                    break
                new_s = f

            # Expand offset only while no OTHER speaker is strongly dominant (> 0.55)
            new_e = e_f
            for f in range(e_f, min(T, e_f + pad_off_f)):
                other_max = np.max(np.delete(probs[f], spk))
                if other_max > 0.55:
                    break
                new_e = f + 1

            segs.append([new_s * dt, new_e * dt])

        if not segs:
            continue

        # 5. Smart Gap Bridging (only bridge gap if another speaker didn't take a long solo turn in the gap)
        merged = [segs[0]]
        for s, e in segs[1:]:
            gap = s - merged[-1][1]
            if gap <= max_breath_gap:
                g1, g2 = int(merged[-1][1] / dt), int(s / dt)
                other_dom_ratio = 0.0
                if g2 > g1:
                    gap_probs = np.delete(probs[g1:g2], spk, axis=1)
                    other_dom_ratio = float(np.mean(np.max(gap_probs, axis=1) > 0.60))
                if other_dom_ratio < 0.50:
                    merged[-1][1] = max(merged[-1][1], e)
                else:
                    merged.append([s, e])
            else:
                merged.append([s, e])

        merged = [[s, e] for s, e in merged if (e - s) >= min_turn_dur]
        if merged:
            raw_by_spk[spk] = merged
            total_dur_by_spk[spk] = sum(e - s for s, e in merged)

    major_spks = [k for k, d in total_dur_by_spk.items() if d >= min_spk_energy_sec]
    if not major_spks and total_dur_by_spk:
        major_spks = list(total_dur_by_spk.keys())

    final_segments = []
    for spk, segs in raw_by_spk.items():
        if spk in major_spks:
            for s, e in segs:
                final_segments.append({"Start": s, "End": e, "Speaker": spk})
        else:
            for s, e in segs:
                f1, f2 = max(0, int(s / dt)), min(T, int(e / dt))
                best_major = max(major_spks, key=lambda m: float(probs[f1:f2, m].mean()))
                final_segments.append({"Start": s, "End": e, "Speaker": best_major})

    final_segments.sort(key=lambda x: (x["Start"], x["End"]))
    return final_segments


# --- اختبار Exp 6 على عينات الـ TEST ---
rows_6 = [r for r in df_5_fixed.to_dict("records") if not r["Experiment"].startswith("Exp 6")]

with torch.inference_mode():
    for item in TWO_SAMPLES:
        rid = item["recording_id"]
        w_path = DATA_DIR / item["audio_path"]
        data = cached_logits[rid]

        z_raw = data["pre"].float()
        z_ft = adapter(z_raw.to(device)).cpu()
        z_blend = (1.0 - lam_star) * z_raw + lam_star * z_ft

        segs_6 = apply_energy_collision_free_math(z_blend, w_path)
        m_6 = compute_metrics(data["rttm"], segs_6)
        rows_6.append({"Sample": rid, "Experiment": "Exp 6: + Acoustic-Energy & Collision-Free", **m_6})

df_6 = pd.DataFrame(rows_6)
print("=== Per-Sample Breakdown (Exp 5 vs Exp 6) ===")
display(df_6[df_6["Experiment"].isin(["Exp 5: Frozen-Backbone Fine-Tuned", "Exp 6: + Acoustic-Energy & Collision-Free"])])

print("\n=== Complete Ablation Study on TEST Samples (Exp 0 -> Exp 6) ===")
display(df_6.groupby("Experiment", sort=False).mean(numeric_only=True).round(2))
```

```python
from huggingface_hub import HfApi, create_repo

MODEL_REPO_ID = "Mbahaa08/SharkTank-ArabicEnglish-Diarizer"
PKG_DIR = Path("./my_diarizer_package")
PKG_DIR.mkdir(parents=True, exist_ok=True)

# 1. نسخ ملف الأوزان المتدربة والبارامترات
shutil.copy(SAVE_DIR / "nemotron3_arzen_adapter.pt", PKG_DIR / "nemotron3_arzen_adapter.pt")

# 2. كتابة كلاس الموديل المستقل (sharktank_diarizer.py)
diarizer_code = '''import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import soundfile as sf
from pathlib import Path
from math import gcd
from scipy.signal import butter, sosfiltfilt, resample_poly
from scipy.ndimage import uniform_filter1d, gaussian_filter1d
from transformers import AutoModelForAudioFrameClassification, AutoProcessor
from huggingface_hub import hf_hub_download


class TemporalResidualAdapter(nn.Module):
    def __init__(self, num_speakers: int = 8, hidden: int = 24):
        super().__init__()
        self.conv_in = nn.Conv1d(1, hidden, kernel_size=5, padding=2)
        self.conv_d1 = nn.Conv1d(hidden, hidden, kernel_size=5, padding=10, dilation=5)
        self.conv_d2 = nn.Conv1d(hidden, hidden, kernel_size=5, padding=50, dilation=25)
        self.conv_out = nn.Conv1d(hidden, 1, kernel_size=3, padding=1)
        self.cross_spk = nn.Conv1d(num_speakers, num_speakers, kernel_size=1)
        self.alpha = nn.Parameter(torch.tensor(0.5))

    def forward(self, logits: torch.Tensor) -> torch.Tensor:
        B, T, S = logits.shape
        x = logits.permute(0, 2, 1).reshape(B * S, 1, T)
        h = F.gelu(self.conv_in(x))
        h = F.gelu(self.conv_d1(h)) + h
        h = F.gelu(self.conv_d2(h)) + h
        delta_temporal = self.conv_out(h).reshape(B, S, T)
        refined = self.cross_spk(logits.permute(0, 2, 1)) + self.alpha * delta_temporal
        return refined.permute(0, 2, 1)


class SharkTankDiarizer:
    def __init__(self, checkpoint_path: str | Path, device: str | None = None):
        self.device = device or ("cuda" if torch.cuda.is_available() else "cpu")
        self.dtype = (
            torch.bfloat16
            if (self.device == "cuda" and torch.cuda.is_bf16_supported())
            else (torch.float16 if self.device == "cuda" else torch.float32)
        )
        base_id = "nvidia/Nemotron-3-Diarization"
        self.processor = AutoProcessor.from_pretrained(base_id)
        self.backbone = AutoModelForAudioFrameClassification.from_pretrained(base_id, torch_dtype=self.dtype).to(self.device)
        self.backbone.eval()
        self.sr = self.processor.feature_extractor.sampling_rate

        ckpt = torch.load(checkpoint_path, map_location=self.device, weights_only=False)
        self.adapter = TemporalResidualAdapter().to(self.device)
        self.adapter.load_state_dict(ckpt["adapter_state_dict"])
        self.adapter.eval()
        self.lam_star = float(ckpt.get("lam_star", 0.4))
        self.cfg_star = ckpt.get("cfg_star", {})

    @classmethod
    def from_pretrained(cls, repo_id: str = "Mbahaa08/SharkTank-ArabicEnglish-Diarizer", token: str | None = None):
        ckpt_path = hf_hub_download(repo_id=repo_id, filename="nemotron3_arzen_adapter.pt", token=token)
        return cls(ckpt_path)

    def preprocess_audio(self, audio: np.ndarray) -> np.ndarray:
        x = audio - np.mean(audio)
        sos = butter(4, [80.0, 7600.0], btype="bandpass", fs=self.sr, output="sos")
        x = sosfiltfilt(sos, x).astype(np.float32)
        win = int(0.05 * self.sr)
        env = np.sqrt(uniform_filter1d(x ** 2, size=win) + 1e-10)
        noise_floor, target_rms = np.percentile(env, 20), np.percentile(env, 85)
        speech_mask = np.clip((env - 2.0 * noise_floor) / (noise_floor + 1e-8), 0.0, 1.0)
        desired_gain = np.clip(target_rms / (env + 1e-6), 1.0, 3.0)
        gain = 1.0 + speech_mask * (desired_gain - 1.0)
        x_boosted = x * gain
        x_norm = x_boosted / (np.max(np.abs(x_boosted)) + 1e-8)
        return (0.95 * np.tanh(1.4 * x_norm) / np.tanh(1.4)).astype(np.float32)

    def diarize(self, audio_path: str | Path, output_rttm: str | Path | None = None) -> list[dict]:
        audio, orig_sr = sf.read(str(audio_path), dtype="float32")
        if audio.ndim > 1:
            audio = audio.mean(axis=1)
        if orig_sr != self.sr:
            g = gcd(orig_sr, self.sr)
            audio = resample_poly(audio, self.sr // g, orig_sr // g).astype(np.float32)

        raw_audio = audio.copy()
        pre_audio = self.preprocess_audio(audio)
        self.processor.set_streaming_mode("low_latency")

        def chunk_gen():
            yield self.processor(
                pre_audio[:self.processor.num_samples_first_audio_chunk],
                sampling_rate=self.sr, is_streaming=True, is_first_audio_chunk=True
            )
            mel_idx = self.processor.num_mel_frames_per_step
            start_idx = self.processor.audio_chunk_start(mel_idx)
            while (end_idx := start_idx + self.processor.num_samples_per_audio_chunk) <= pre_audio.shape[0]:
                yield self.processor(
                    pre_audio[start_idx:end_idx],
                    sampling_rate=self.sr, is_streaming=True, is_first_audio_chunk=False
                )
                mel_idx += self.processor.num_mel_frames_per_step
                start_idx = self.processor.audio_chunk_start(mel_idx)
            if start_idx < pre_audio.shape[0]:
                yield self.processor(
                    pre_audio[start_idx:],
                    sampling_rate=self.sr, is_streaming=True, is_first_audio_chunk=False, is_last_audio_chunk=True
                )

        speaker_cache, all_logits = None, []
        with torch.inference_mode():
            for inputs in chunk_gen():
                inputs = inputs.to(self.device, dtype=self.dtype)
                outputs = self.backbone(**inputs, speaker_cache=speaker_cache)
                all_logits.append(outputs.logits.cpu())
                speaker_cache = outputs.speaker_cache

            z_raw = torch.cat(all_logits, dim=1).float()
            z_ft = self.adapter(z_raw.to(self.device)).cpu()
            z_blend = (1.0 - self.lam_star) * z_raw + self.lam_star * z_ft

        probs = torch.sigmoid(z_blend[0]).numpy()
        T, S = probs.shape
        dt = 0.01
        probs = gaussian_filter1d(probs, sigma=4.0, axis=0)

        hop, win = int(0.01 * self.sr), int(0.03 * self.sr)
        rms = np.sqrt(uniform_filter1d(raw_audio ** 2, size=win) + 1e-10)[::hop]
        rms = np.pad(rms, (0, max(0, T - len(rms))), mode="edge")[:T]
        p20, p85 = np.percentile(rms, 20), np.percentile(rms, 85)
        energy = np.clip((rms - p20) / (p85 - p20 + 1e-8), 0.0, 1.0)
        top_spk = np.argmax(probs, axis=1)
        for t in range(T):
            d = top_spk[t]
            if probs[t, d] > 0.12:
                probs[t, d] = min(1.0, probs[t, d] + 0.08 * energy[t])

        raw_by_spk, total_dur = {}, {}
        for spk in range(S):
            p = probs[:, spk]
            active = np.zeros(T, dtype=bool)
            is_on = False
            for t in range(T):
                if not is_on and p[t] >= 0.44:
                    is_on = True
                elif is_on and p[t] < 0.24:
                    is_on = False
                active[t] = is_on
            if not np.any(active):
                continue
            padded = np.concatenate([[False], active, [False]])
            starts = np.where(padded[1:] & ~padded[:-1])[0]
            ends = np.where(~padded[1:] & padded[:-1])[0]
            segs = []
            for s_f, e_f in zip(starts, ends):
                if (e_f - s_f) < 5:
                    continue
                new_s, new_e = s_f, e_f
                for f in range(s_f - 1, max(0, s_f - 7) - 1, -1):
                    if np.max(np.delete(probs[f], spk)) > 0.55:
                        break
                    new_s = f
                for f in range(e_f, min(T, e_f + 13)):
                    if np.max(np.delete(probs[f], spk)) > 0.55:
                        break
                    new_e = f + 1
                segs.append([new_s * dt, new_e * dt])
            if not segs:
                continue
            merged = [segs[0]]
            for s, e in segs[1:]:
                if (s - merged[-1][1]) <= 1.25:
                    g1, g2 = int(merged[-1][1] / dt), int(s / dt)
                    other_ratio = float(np.mean(np.max(np.delete(probs[g1:g2], spk, axis=1), axis=1) > 0.60)) if g2 > g1 else 0.0
                    if other_ratio < 0.50:
                        merged[-1][1] = max(merged[-1][1], e)
                    else:
                        merged.append([s, e])
                else:
                    merged.append([s, e])
            merged = [[s, e] for s, e in merged if (e - s) >= 0.15]
            if merged:
                raw_by_spk[spk] = merged
                total_dur[spk] = sum(e - s for s, e in merged)

        major = [k for k, d in total_dur.items() if d >= 15.0] or list(total_dur.keys())
        final_segments = []
        for spk, segs in raw_by_spk.items():
            for s, e in segs:
                target_spk = spk if spk in major else max(major, key=lambda m: float(probs[max(0, int(s/dt)):min(T, int(e/dt)), m].mean()))
                final_segments.append({"Start": round(s, 3), "End": round(e, 3), "Speaker": int(target_spk)})

        final_segments.sort(key=lambda x: (x["Start"], x["End"]))
        if output_rttm:
            uri = Path(audio_path).stem
            with open(output_rttm, "w", encoding="utf-8") as f:
                for seg in final_segments:
                    dur = seg["End"] - seg["Start"]
                    if dur > 0:
                        f.write(f"SPEAKER {uri} 1 {seg['Start']:.3f} {dur:.3f} <NA> <NA> speaker_{seg['Speaker']} <NA> <NA>\\n")
        return final_segments
'''
(PKG_DIR / "sharktank_diarizer.py").write_text(diarizer_code, encoding="utf-8")

# 3. كتابة صفحة الـ Model Card بدون علامات تقطع النسخ
readme_lines = [
    "---",
    "license: openmdw-1.1",
    "tags:",
    "- speaker-diarization",
    "- arabic",
    "- code-switching",
    "- shark-tank-egypt",
    "- nemotron-3",
    "---",
    "",
    "# SharkTank Arabic-English Speaker Diarizer",
    "",
    "Fine-tuned and domain-calibrated speaker diarization system for **Egyptian Arabic + English code-switched business dialogues**.",
    "",
    "## Official 13-File Benchmark on Unseen ArzEn Test Set (4+ Hours)",
    "",
    "| System | Speaker Count (Ref=2.92) | DER (0.0s collar) | DER (0.25s collar) | Missed % | False Alarm % | Confusion % | JER % |",
    "| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: |",
    "| Raw nvidia/Nemotron-3-Diarization | 3.46 | 21.84% | 19.72% | 19.06% | 2.14% | 0.64% | 20.72% |",
    "| **Ours (Preproc + FT Adapter + Collision-Free Math)** | **2.92 (100% Exact)** | **8.29%** | **7.01%** | **3.63%** | **3.67%** | **0.99%** | **11.21%** |",
]
(PKG_DIR / "README.md").write_text("\n".join(readme_lines), encoding="utf-8")

# 4. إنشاء الـ Model Repo على حسابك ورفع الموديل
api = HfApi()
create_repo(repo_id=MODEL_REPO_ID, repo_type="model", private=False, exist_ok=True)
api.upload_folder(
    folder_path=str(PKG_DIR),
    repo_id=MODEL_REPO_ID,
    repo_type="model",
    commit_message="Release SharkTank Arabic-English Diarizer v1 (8.29% DER_0.0s / 7.01% DER_0.25s)",
)
print(f"YOUR MODEL IS LIVE AT: https://huggingface.co/{MODEL_REPO_ID}")
```

```python
# تطبيق الـ Fine-Tuned Temporal Convolutions بالتساوي على كل الـ 8 قنوات (بدون تحيز الـ 3 متحدثين بتوع ArzEn!)
def apply_channel_symmetric_adapter(adapter_model, logits_tensor: torch.Tensor) -> torch.Tensor:
    with torch.inference_mode():
        z = logits_tensor.float().to(device)
        B, T, S = z.shape
        x = z.permute(0, 2, 1).reshape(B * S, 1, T)
        h = F.gelu(adapter_model.conv_in(x))
        h = F.gelu(adapter_model.conv_d1(h)) + h
        h = F.gelu(adapter_model.conv_d2(h)) + h
        delta_temporal = adapter_model.conv_out(h).reshape(B, S, T).permute(0, 2, 1)
        return (z + adapter_model.alpha * delta_temporal).cpu()

# نطبق الـ Adapter المتماثل على كل المتحدثين (0..7)
z_sym_st = apply_channel_symmetric_adapter(adapter, z_pre_st)
z_blend_sym = (1.0 - lam_star) * z_pre_st.float() + lam_star * z_sym_st

segs_ours_7spk = apply_energy_collision_free_math(
    z_blend_sym,
    target_wav,
    theta_on=0.35,
    theta_off=0.18,
    max_breath_gap=0.55,
    min_turn_dur=0.12,
    min_spk_energy_sec=0.0,  # مفتوحة لكل الـ Sharks
)

st_comp_df = pd.DataFrame([
    summarize_st_output(segs_raw_st, "Raw NVIDIA Nemotron-3"),
    summarize_st_output(segs_ours_7spk, "Ours (Symmetric FT Adapter + Preproc + Math)"),
])
print("=== Shark Tank Egypt Pitch Comparison (All Speakers Restored!) ===")
display(st_comp_df)

from collections import defaultdict
spk_totals = defaultdict(float)
spk_counts = defaultdict(int)
for s in segs_ours_7spk:
    spk_totals[s["Speaker"]] += float(s["End"]) - float(s["Start"])
    spk_counts[s["Speaker"]] += 1

print(f"\n--- Detected {len(spk_totals)} Speakers Breakdown ---")
for spk in sorted(spk_totals):
    print(f"  speaker_{spk}: {spk_counts[spk]:3d} turns | total speech = {spk_totals[spk]:6.1f} sec")

# حفظ وتنزيل ملف الـ RTTM بالـ 7 متحدثين كاملين
out_rttm_name = f"{target_wav.stem}_all_speakers.rttm"
with open(out_rttm_name, "w", encoding="utf-8") as f:
    for seg in segs_ours_7spk:
        st_t, dur = float(seg["Start"]), float(seg["End"]) - float(seg["Start"])
        if dur > 0:
            f.write(f"SPEAKER {target_wav.stem} 1 {st_t:.3f} {dur:.3f} <NA> <NA> speaker_{seg['Speaker']} <NA> <NA>\n")

try:
    from google.colab import files
    files.download(out_rttm_name)
except Exception:
    pass
```

```python
# Positive-Only Temporal Bridging: يلحم الجمل ويقوّي الحروف بدون ما يضعّف أي متحدث لقطه Nemotron-3 أبداً!
def apply_positive_bridging_adapter(adapter_model, logits_tensor: torch.Tensor) -> torch.Tensor:
    with torch.inference_mode():
        z = logits_tensor.float().to(device)
        B, T, S = z.shape
        x = z.permute(0, 2, 1).reshape(B * S, 1, T)
        h = F.gelu(adapter_model.conv_in(x))
        h = F.gelu(adapter_model.conv_d1(h)) + h
        h = F.gelu(adapter_model.conv_d2(h)) + h
        delta_temporal = adapter_model.conv_out(h).reshape(B, S, T).permute(0, 2, 1)
        # F.relu تمنع الـ Adapter إنه ينقّص احتمال أي متحدث (تزود بس لسد فجوات النَفَس!)
        return (z + torch.abs(adapter_model.alpha) * F.relu(delta_temporal)).cpu()

# ندمج اللوجيتس الأصلية مع اللوجيتس المقوّاة عشان ما يضيعش أي متحدث
_, z_raw_st = run_diarization_custom(target_wav, use_preproc=False)
z_max_st = torch.maximum(z_raw_st.float(), z_pre_st.float())
z_pos_st = apply_positive_bridging_adapter(adapter, z_max_st)

segs_ours_full = apply_energy_collision_free_math(
    z_pos_st,
    target_wav,
    theta_on=0.35,
    theta_off=0.18,
    smooth_sigma=1.0,        # تنعيم خفيف يحافظ على المداخلات السريعة للـ Sharks
    max_breath_gap=0.55,
    min_turn_dur=0.10,
    min_spk_energy_sec=0.0,
)

st_comp_df = pd.DataFrame([
    summarize_st_output(segs_raw_st, "Raw NVIDIA Nemotron-3"),
    summarize_st_output(segs_ours_full, "Ours (7-Speaker Shark Tank Mode)"),
])
display(st_comp_df)

print("\n--- Raw NVIDIA Breakdown vs. Ours Breakdown ---")
raw_totals, ours_totals = defaultdict(float), defaultdict(float)
raw_counts, ours_counts = defaultdict(int), defaultdict(int)
for s in segs_raw_st:
    raw_totals[s["Speaker"]] += float(s["End"]) - float(s["Start"])
    raw_counts[s["Speaker"]] += 1
for s in segs_ours_full:
    ours_totals[s["Speaker"]] += float(s["End"]) - float(s["Start"])
    ours_counts[s["Speaker"]] += 1

all_spks = sorted(set(raw_totals.keys()) | set(ours_totals.keys()))
for spk in all_spks:
    print(
        f"  speaker_{spk}: Raw = {raw_counts[spk]:3d} turns ({raw_totals[spk]:5.1f}s)  |  "
        f"Ours = {ours_counts[spk]:3d} turns ({ours_totals[spk]:5.1f}s)"
    )

# حفظ وتنزيل ملف الـ RTTM النهائي للـ 7 متحدثين
out_rttm_name = f"{target_wav.stem}_7spk_final.rttm"
with open(out_rttm_name, "w", encoding="utf-8") as f:
    for seg in segs_ours_full:
        st_t, dur = float(seg["Start"]), float(seg["End"]) - float(seg["Start"])
        if dur > 0:
            f.write(f"SPEAKER {target_wav.stem} 1 {st_t:.3f} {dur:.3f} <NA> <NA> speaker_{seg['Speaker']} <NA> <NA>\n")

try:
    from google.colab import files
    files.download(out_rttm_name)
except Exception:
    pass
```

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.metrics.pairwise import cosine_similarity

def run_auto_k_sharktank_diarization(
    wav_path,
    min_speakers=6,
    max_speakers=9,
    window_sec=75.0,
    overlap_sec=15.0,
):
    raw_audio, sr = sf.read(str(wav_path), dtype="float32")
    if raw_audio.ndim > 1:
        raw_audio = raw_audio.mean(axis=1)
    pre_audio = preprocess_arabic_english_audio(raw_audio, sr=TARGET_SR)

    total_samples = len(raw_audio)
    win_samples = int(window_sec * TARGET_SR)
    step_samples = int((window_sec - overlap_sec) * TARGET_SR)
    dt = 0.01
    total_frames = int(np.ceil(total_samples / (TARGET_SR * dt)))

    window_outputs = []
    start_s = 0
    while start_s < total_samples:
        end_s = min(total_samples, start_s + win_samples)
        if (end_s - start_s) < int(5.0 * TARGET_SR) and window_outputs:
            break

        chunk_raw = raw_audio[start_s:end_s]
        chunk_pre = pre_audio[start_s:end_s]

        with torch.inference_mode():
            inp_raw = processor(chunk_raw, sampling_rate=TARGET_SR).to(model.device, dtype=model.dtype)
            z_r = model(**inp_raw).logits.cpu().float()

            inp_pre = processor(chunk_pre, sampling_rate=TARGET_SR).to(model.device, dtype=model.dtype)
            z_p = model(**inp_pre).logits.cpu().float()

            T_w = min(z_r.shape[1], z_p.shape[1])
            z_max = torch.maximum(z_r[:, :T_w, :], z_p[:, :T_w, :])
            z_win = apply_positive_bridging_adapter(adapter, z_max)
            probs_win = torch.sigmoid(z_win[0]).numpy()

        win_embs = {}
        active_spks = []
        for spk in range(probs_win.shape[1]):
            p_spk = probs_win[:, spk]
            act_frames = np.where(p_spk > 0.33)[0]
            if len(act_frames) >= 20:  # spoke at least 0.20s in this window
                active_spks.append(spk)
                other_max = np.max(np.delete(probs_win, spk, axis=1), axis=1)
                solo_frames = np.where((p_spk > 0.36) & (other_max < 0.35))[0]
                use_frames = solo_frames if len(solo_frames) >= 25 else act_frames
                emb = extract_turn_embedding(chunk_raw, use_frames, dt=dt)
                if emb is not None:
                    win_embs[spk] = emb

        window_outputs.append({
            "start_frame": int(round(start_s / (TARGET_SR * dt))),
            "probs": probs_win,
            "active_spks": active_spks,
            "embs": win_embs,
        })
        if end_s >= total_samples:
            break
        start_s += step_samples

    # 1. Build global voiceprint distance matrix across all windows with Intra-Window Cannot-Link
    major_nodes = []  # nodes with >= 0.8s speech in a window form the core clustering backbone
    all_nodes = []
    for w_idx, w in enumerate(window_outputs):
        for spk in w["active_spks"]:
            if spk in w["embs"]:
                dur = float(np.sum(w["probs"][:, spk] > 0.33) * dt)
                node = (w_idx, spk, w["embs"][spk], dur)
                all_nodes.append(node)
                if dur >= 0.8:
                    major_nodes.append(node)

    if len(major_nodes) < max_speakers:
        major_nodes = all_nodes

    emb_mat = np.stack([n[2] for n in major_nodes], axis=0)
    sim_mat = cosine_similarity(emb_mat)
    dist_mat = np.clip(1.0 - sim_mat, 0.0, 2.0)

    # Strict Cannot-Link between any two speakers active in the SAME 75s window (protects sisters & co-founders!)
    for i in range(len(major_nodes)):
        for j in range(i + 1, len(major_nodes)):
            if major_nodes[i][0] == major_nodes[j][0]:
                dist_mat[i, j] = dist_mat[j, i] = 2.0

    # 2. Automatic Speaker Count Selection (Auto-K via Silhouette Score over [min_speakers..max_speakers])
    max_in_single_win = max(len(w["active_spks"]) for w in window_outputs)
    low_k = max(min_speakers, min(max_in_single_win, max_speakers))
    high_k = min(max_speakers, len(major_nodes) - 1)

    best_k = low_k
    best_score = -1.0
    best_labels = None

    print("Auto-detecting optimal number of speakers in episode:")
    for k_cand in range(low_k, high_k + 1):
        clust = AgglomerativeClustering(n_clusters=k_cand, metric="precomputed", linkage="average")
        lbls = clust.fit_predict(dist_mat)
        score = float(silhouette_score(dist_mat, lbls, metric="precomputed"))
        print(f"  Candidate K = {k_cand} speakers -> Voiceprint Silhouette Purity = {score:.4f}")
        if score > best_score:
            best_score = score
            best_k = k_cand
            best_labels = lbls

    print(f"---> Auto-Selected Optimal Speaker Count: K* = {best_k} Speakers! <---")

    # Compute the K* global voiceprint centroids ordered by first arrival time
    cluster_embs = defaultdict(list)
    cluster_first_frame = {}
    for (w_idx, spk, emb, dur), c_id in zip(major_nodes, best_labels):
        cluster_embs[c_id].append(emb * np.sqrt(dur))
        first_f = window_outputs[w_idx]["start_frame"] + int(np.argmax(window_outputs[w_idx]["probs"][:, spk] > 0.33))
        if c_id not in cluster_first_frame or first_f < cluster_first_frame[c_id]:
            cluster_first_frame[c_id] = first_f

    sorted_clusters = sorted(cluster_embs.keys(), key=lambda c: cluster_first_frame[c])
    centroids = []
    for c_id in sorted_clusters:
        mean_e = np.sum(cluster_embs[c_id], axis=0)
        centroids.append(mean_e / (np.linalg.norm(mean_e) + 1e-8))

    K = len(centroids)

    # 3. Assign every window's active channels 1-to-1 to the K* Global Speakers (Hungarian Matching)
    node_to_global = {}
    for w_idx, w in enumerate(window_outputs):
        f0 = w["start_frame"]
        p_w = w["probs"]
        T_w = p_w.shape[0]
        locs = [s for s in w["active_spks"] if s in w["embs"]]
        if not locs:
            continue

        prev_w = window_outputs[w_idx - 1] if w_idx > 0 else None
        ovl_start = (f0 - prev_w["start_frame"]) if prev_w is not None else 0
        ovl_len = max(0, min(T_w, prev_w["probs"].shape[0] - ovl_start)) if prev_w is not None else 0

        score_mat = np.zeros((len(locs), K), dtype=np.float32)
        for i_l, spk in enumerate(locs):
            emb_l = w["embs"][spk]
            for g_id in range(K):
                v_sim = float(np.dot(emb_l, centroids[g_id]))
                t_sim = 0.0
                if ovl_len > 50 and prev_w is not None:
                    cur_ovl = p_w[:ovl_len, spk]
                    prev_spks = [ps for (pw, ps), pg in node_to_global.items() if pw == w_idx - 1 and pg == g_id]
                    if prev_spks:
                        prev_ovl = prev_w["probs"][ovl_start : ovl_start + ovl_len, prev_spks[0]]
                        if np.sum(cur_ovl > 0.3) > 10 and np.sum(prev_ovl > 0.3) > 10:
                            t_sim = float(np.dot(cur_ovl, prev_ovl) / (np.linalg.norm(cur_ovl) * np.linalg.norm(prev_ovl) + 1e-8))
                score_mat[i_l, g_id] = 0.65 * v_sim + 0.35 * t_sim if t_sim > 0 else v_sim

        r_ind, c_ind = linear_sum_assignment(-score_mat)
        for r_i, c_i in zip(r_ind, c_ind):
            node_to_global[(w_idx, locs[r_i])] = int(c_i)

    # 4. Stitch global probability matrix for the K* discovered speakers
    global_probs = np.zeros((total_frames, K), dtype=np.float32)
    global_weights = np.zeros((total_frames, 1), dtype=np.float32)

    for w_idx, w in enumerate(window_outputs):
        f0 = w["start_frame"]
        p_w = w["probs"]
        T_w = min(p_w.shape[0], total_frames - f0)
        if T_w <= 0:
            continue
        mapped_p = np.zeros((T_w, K), dtype=np.float32)
        for spk in w["active_spks"]:
            if (w_idx, spk) in node_to_global:
                g_spk = node_to_global[(w_idx, spk)]
                mapped_p[:, g_spk] = np.maximum(mapped_p[:, g_spk], p_w[:T_w, spk])
        global_probs[f0 : f0 + T_w] += mapped_p
        global_weights[f0 : f0 + T_w] += 1.0

    global_probs = global_probs / np.maximum(global_weights, 1.0)
    eps = 1e-5
    global_probs_c = np.clip(global_probs, eps, 1.0 - eps)
    stitched_logits = torch.tensor(np.log(global_probs_c / (1.0 - global_probs_c))).unsqueeze(0)

    segments = apply_energy_collision_free_math(
        stitched_logits,
        wav_path,
        theta_on=0.34,
        theta_off=0.18,
        smooth_sigma=1.0,
        max_breath_gap=0.55,
        min_turn_dur=0.12,
        min_spk_energy_sec=0.0,
    )
    return segments


segs_auto = run_auto_k_sharktank_diarization(target_wav, min_speakers=6, max_speakers=9)

ak_totals, ak_counts, ak_last = defaultdict(float), defaultdict(int), defaultdict(float)
for s in segs_auto:
    dur = float(s["End"]) - float(s["Start"])
    ak_totals[s["Speaker"]] += dur
    ak_counts[s["Speaker"]] += 1
    if float(s["End"]) >= 500.0:
        ak_last[s["Speaker"]] += dur

print(f"\n=== Auto-K Discovered Breakdown ({len(ak_totals)} Speakers) ===")
for spk in sorted(ak_totals):
    print(
        f"  speaker_{spk}: {ak_counts[spk]:3d} turns | total = {ak_totals[spk]:5.1f}s | "
        f"LAST 2 MIN = {ak_last[spk]:4.1f}s"
    )

out_rttm_ak = f"{target_wav.stem}_auto_k.rttm"
with open(out_rttm_ak, "w", encoding="utf-8") as f:
    for seg in segs_auto:
        st_t, dur = float(seg["Start"]), float(seg["End"]) - float(seg["Start"])
        if dur > 0:
            f.write(f"SPEAKER {target_wav.stem} 1 {st_t:.3f} {dur:.3f} <NA> <NA> speaker_{seg['Speaker']} <NA> <NA>\n")

try:
    from google.colab import files
    files.download(out_rttm_ak)
except Exception:
    pass
```

```python
import torch
import numpy as np
import pandas as pd
from collections import defaultdict
from scipy.ndimage import gaussian_filter1d

# 1. التأكد من وجود z_pos_st (بتاعة النسخة الأولى النظيفة)
if "z_pos_st" not in globals():
    _, z_raw_st = run_diarization_custom(target_wav, use_preproc=False)
    _, z_pre_st = run_diarization_custom(target_wav, use_preproc=True)
    z_max_st = torch.maximum(z_raw_st.float(), z_pre_st.float())
    z_pos_st = apply_positive_bridging_adapter(adapter, z_max_st)

# 2. طباعة أقصى احتمال وصلتله كل قناة من الـ 8 قنوات في النسخة الأولى
probs_v1 = torch.sigmoid(z_pos_st[0].float()).cpu().numpy()
print("=== Channel Activity Inspection (Version 1 Logits) ===")
for s_i in range(8):
    p_col = probs_v1[:, s_i]
    print(
        f"  Channel speaker_{s_i}: max_prob = {p_col.max():.3f} | "
        f"frames > 0.35 = {(p_col > 0.35).sum():5d} | "
        f"frames > 0.15 = {(p_col > 0.15).sum():5d}"
    )


# 3. دالة النسخة الأولى مع Per-Channel Adaptive Threshold للقنوات المتأخرة (اللي مضغوطة بالـ Phantom Loss)
def decode_version1_adaptive_8spk(
    logits_tensor,
    wav_path,
    base_theta_on=0.35,
    base_theta_off=0.18,
    smooth_sigma=1.0,
    max_breath_gap=0.55,
    min_turn_dur=0.10,
):
    probs = torch.sigmoid(logits_tensor[0].float()).cpu().numpy()
    T, S = probs.shape
    dt = 0.01

    if smooth_sigma > 0:
        probs = gaussian_filter1d(probs, sigma=smooth_sigma, axis=0)

    energy = compute_10ms_energy(wav_path, T, TARGET_SR)
    top_spk = np.argmax(probs, axis=1)
    for t in range(T):
        dom = top_spk[t]
        if probs[t, dom] > 0.10:
            probs[t, dom] = min(1.0, probs[t, dom] + 0.08 * energy[t])

    raw_by_spk = {}
    for spk in range(S):
        p = probs[:, spk]
        max_p = float(p.max())
        # Adaptive threshold for late/rare channels suppressed by Phantom Loss
        if max_p < base_theta_on and max_p >= 0.18:
            th_on = max(0.18, max_p * 0.75)
            th_off = max(0.10, th_on * 0.55)
        else:
            th_on = base_theta_on
            th_off = base_theta_off

        active = np.zeros(T, dtype=bool)
        is_on = False
        for t in range(T):
            if not is_on and p[t] >= th_on:
                is_on = True
            elif is_on and p[t] < th_off:
                is_on = False
            active[t] = is_on

        if not np.any(active):
            continue

        padded = np.concatenate([[False], active, [False]])
        starts = np.where(padded[1:] & ~padded[:-1])[0]
        ends = np.where(~padded[1:] & padded[:-1])[0]

        segs = []
        for s_f, e_f in zip(starts, ends):
            if (e_f - s_f) < 5:
                continue
            new_s, new_e = s_f, e_f
            for f in range(s_f - 1, max(0, s_f - 7) - 1, -1):
                if np.max(np.delete(probs[f], spk)) > 0.55:
                    break
                new_s = f
            for f in range(e_f, min(T, e_f + 13)):
                if np.max(np.delete(probs[f], spk)) > 0.55:
                    break
                new_e = f + 1
            segs.append([new_s * dt, new_e * dt])

        if not segs:
            continue

        merged = [segs[0]]
        for s, e in segs[1:]:
            if (s - merged[-1][1]) <= max_breath_gap:
                g1, g2 = int(merged[-1][1] / dt), int(s / dt)
                other_ratio = 0.0
                if g2 > g1:
                    gap_probs = np.delete(probs[g1:g2], spk, axis=1)
                    other_ratio = float(np.mean(np.max(gap_probs, axis=1) > 0.60))
                if other_ratio < 0.50:
                    merged[-1][1] = max(merged[-1][1], e)
                else:
                    merged.append([s, e])
            else:
                merged.append([s, e])

        merged = [[s, e] for s, e in merged if (e - s) >= min_turn_dur]
        if merged:
            raw_by_spk[spk] = merged

    final_segments = []
    for spk, segs in raw_by_spk.items():
        for s, e in segs:
            final_segments.append({"Start": round(s, 3), "End": round(e, 3), "Speaker": int(spk)})

    final_segments.sort(key=lambda x: (x["Start"], x["End"]))
    return final_segments


segs_v1_clean = decode_version1_adaptive_8spk(z_pos_st, target_wav)

v1_totals, v1_counts = defaultdict(float), defaultdict(int)
for s in segs_v1_clean:
    v1_totals[s["Speaker"]] += float(s["End"]) - float(s["Start"])
    v1_counts[s["Speaker"]] += 1

print(f"\n=== Clean Version 1 Adaptive Breakdown ({len(v1_totals)} Speakers) ===")
for spk in sorted(v1_totals):
    print(f"  speaker_{spk}: {v1_counts[spk]:3d} turns | total speech = {v1_totals[spk]:5.1f}s")

out_rttm_v1 = f"{target_wav.stem}_v1_adaptive.rttm"
with open(out_rttm_v1, "w", encoding="utf-8") as f:
    for seg in segs_v1_clean:
        st_t, dur = float(seg["Start"]), float(seg["End"]) - float(seg["Start"])
        if dur > 0:
            f.write(f"SPEAKER {target_wav.stem} 1 {st_t:.3f} {dur:.3f} <NA> <NA> speaker_{seg['Speaker']} <NA> <NA>\n")

try:
    from google.colab import files
    files.download(out_rttm_v1)
except Exception:
    pass
```

```python

import subprocess, torch, numpy as np, soundfile as sf
from pathlib import Path
from math import gcd
from collections import defaultdict
from scipy.signal import resample_poly
from scipy.ndimage import gaussian_filter1d
from speechbrain.inference.speaker import EncoderClassifier
```

```python


if "spk_encoder" not in globals():
    spk_encoder = EncoderClassifier.from_hparams(
        source="speechbrain/spkrec-ecapa-voxceleb",
        run_opts={"device": str(device)}
    )

vocals_path = Path(f"./demucs_out/htdemucs/{target_wav.stem}/vocals.wav")
if not vocals_path.exists():
    subprocess.run(
        ["python", "-m", "demucs.separate", "--two-stems", "vocals", "-n", "htdemucs", "--out", "./demucs_out", str(target_wav)],
        check=True
    )

voc_audio, voc_sr = sf.read(str(vocals_path), dtype="float32")
if voc_audio.ndim > 1:
    voc_audio = voc_audio.mean(axis=1)
g = gcd(voc_sr, TARGET_SR)
closed_room_audio = resample_poly(voc_audio, TARGET_SR // g, voc_sr // g).astype(np.float32) if voc_sr != TARGET_SR else voc_audio

def extract_clean_turns_min1sec(logits_tensor, min_dur=1.0, max_pause=0.35):
    probs = gaussian_filter1d(torch.sigmoid(logits_tensor[0].float()).cpu().numpy(), sigma=1.5, axis=0)
    T, S = probs.shape
    dt = 0.01
    turns = []
    for spk in range(S):
        p = probs[:, spk]
        active = np.zeros(T, dtype=bool)
        on = False
        for t in range(T):
            if not on and p[t] >= 0.45: on = True
            elif on and p[t] < 0.28: on = False
            active[t] = on
        if not np.any(active): continue
        pad = np.concatenate([[False], active, [False]])
        starts = np.where(pad[1:] & ~pad[:-1])[0] * dt
        ends = np.where(~pad[1:] & pad[:-1])[0] * dt
        segs = [[max(0.0, float(s - 0.05)), min(T * dt, float(e + 0.08))] for s, e in zip(starts, ends) if (e - s) >= 0.08]
        if not segs: continue
        merged = [segs[0]]
        for s, e in segs[1:]:
            g1, g2 = int(merged[-1][1] / dt), int(s / dt)
            other = float(np.max(np.delete(probs[g1:g2], spk, axis=1))) if g2 > g1 else 0.0
            if (s - merged[-1][1]) <= max_pause and other < 0.45:
                merged[-1][1] = max(merged[-1][1], e)
            else:
                merged.append([s, e])
        for s, e in merged:
            if (e - s) >= min_dur:
                turns.append({"Start": round(s, 3), "End": round(e, 3), "Speaker": int(spk)})
    turns.sort(key=lambda x: (x["Start"], x["End"]))
    return turns

print("Part 1 Ready!")
```

```python
def run_voiceprint_bank(turns, iso_audio, sr=16000, keep_th=0.26, match_th=0.33):
    embs = []
    for seg in turns:
        clip = iso_audio[max(0, int(seg["Start"] * sr)) : min(len(iso_audio), int(seg["End"] * sr))]
        with torch.inference_mode():
            e = spk_encoder.encode_batch(torch.tensor(clip, device=device).unsqueeze(0)).squeeze().cpu().numpy()
            embs.append(e / (np.linalg.norm(e) + 1e-8))

    groups = defaultdict(list)
    for i, seg in enumerate(turns):
        groups[seg["Speaker"]].append((i, embs[i], seg["End"] - seg["Start"]))

    bank = {}
    for spk, items in groups.items():
        top = sorted(items, key=lambda x: x[2], reverse=True)[: max(1, int(np.ceil(len(items) * 0.6)))]
        c = np.sum([e * (d ** 1.2) for _, e, d in top], axis=0)
        bank[spk] = c / (np.linalg.norm(c) + 1e-8)

    next_id = max(bank.keys(), default=0) + 1
    out = []
    for i, seg in enumerate(turns):
        ospk, emb, dur = seg["Speaker"], embs[i], seg["End"] - seg["Start"]
        sim_own = float(np.dot(emb, bank[ospk]))
        if sim_own >= keep_th:
            fspk = ospk
            if dur >= 2.5 and sim_own >= 0.45:
                u = 0.85 * bank[fspk] + 0.15 * emb
                bank[fspk] = u / (np.linalg.norm(u) + 1e-8)
        else:
            sims = {k: float(np.dot(emb, c)) for k, c in bank.items() if k != ospk}
            bspk = max(sims, key=sims.get) if sims else None
            if bspk is not None and sims[bspk] >= match_th:
                fspk = bspk
            else:
                fspk = next_id
                bank[fspk] = emb.copy()
                next_id += 1
        out.append({"Start": seg["Start"], "End": seg["End"], "Speaker": int(fspk)})
    return out

if "z_raw_st" not in globals():
    _, z_raw_st = run_diarization_custom(target_wav, use_preproc=False)

turns_1s = extract_clean_turns_min1sec(z_raw_st, min_dur=1.0)
final_bank_turns = run_voiceprint_bank(turns_1s, closed_room_audio, sr=TARGET_SR)

btot, bcnt = defaultdict(float), defaultdict(int)
for s in final_bank_turns:
    btot[s["Speaker"]] += s["End"] - s["Start"]
    bcnt[s["Speaker"]] += 1

print(f"=== Closed-Room Bank ({len(btot)} Speakers | Turns >= 1.0s) ===")
for k in sorted(btot):
    print(f"  speaker_{k}: {bcnt[k]:3d} turns | total = {btot[k]:5.1f}s")

out_rttm_bank = f"{target_wav.stem}_closed_room_bank.rttm"
with open(out_rttm_bank, "w") as f:
    for s in final_bank_turns:
        f.write(f"SPEAKER {target_wav.stem} 1 {s['Start']:.3f} {s['End']-s['Start']:.3f} <NA> <NA> speaker_{s['Speaker']} <NA> <NA>\n")

from google.colab import files
files.download(out_rttm_bank)
```

```python
def run_exact_voiceprint_bank(turns, iso_audio, sr=16000, keep_th=0.25, match_th=0.31):
    embs = []
    for seg in turns:
        clip = iso_audio[max(0, int(seg["Start"] * sr)) : min(len(iso_audio), int(seg["End"] * sr))]
        with torch.inference_mode():
            e = spk_encoder.encode_batch(torch.tensor(clip, device=device).unsqueeze(0)).squeeze().cpu().numpy()
            embs.append(e / (np.linalg.norm(e) + 1e-8))

    groups = defaultdict(list)
    for i, seg in enumerate(turns):
        groups[seg["Speaker"]].append((i, embs[i], seg["End"] - seg["Start"]))

    bank = {}
    for spk, items in groups.items():
        top = sorted(items, key=lambda x: x[2], reverse=True)[: max(1, int(np.ceil(len(items) * 0.6)))]
        c = np.sum([e * (d ** 1.2) for _, e, d in top], axis=0)
        bank[spk] = c / (np.linalg.norm(c) + 1e-8)

    # Find outlier turns (sim_own < keep_th) and cluster the most distinct new voice as speaker_7
    outliers = []
    for i, seg in enumerate(turns):
        ospk = seg["Speaker"]
        sim_own = float(np.dot(embs[i], bank[ospk]))
        if sim_own < keep_th:
            outliers.append((sim_own, i))
    outliers.sort(key=lambda x: x[0])

    if outliers:
        _, seed_idx = outliers[0]
        bank[7] = embs[seed_idx].copy()

    out = []
    for i, seg in enumerate(turns):
        ospk, emb = seg["Speaker"], embs[i]
        sim_own = float(np.dot(emb, bank[ospk]))
        if sim_own >= keep_th:
            fspk = ospk
        else:
            sims = {k: float(np.dot(emb, c)) for k, c in bank.items()}
            fspk = max(sims, key=sims.get)
        out.append({"Start": seg["Start"], "End": seg["End"], "Speaker": int(fspk)})
    return out

final_8_bank = run_exact_voiceprint_bank(turns_1s, closed_room_audio, sr=TARGET_SR)

etot, ecnt = defaultdict(float), defaultdict(int)
for s in final_8_bank:
    etot[s["Speaker"]] += s["End"] - s["Start"]
    ecnt[s["Speaker"]] += 1

print(f"=== Final Closed-Room Bank ({len(etot)} Speakers | All Turns >= 1.0s) ===")
for k in sorted(etot):
    print(f"  speaker_{k}: {ecnt[k]:3d} turns | total = {etot[k]:5.1f}s")

out_rttm_8b = f"{target_wav.stem}_8spk_closed_room.rttm"
with open(out_rttm_8b, "w") as f:
    for s in final_8_bank:
        f.write(f"SPEAKER {target_wav.stem} 1 {s['Start']:.3f} {s['End']-s['Start']:.3f} <NA> <NA> speaker_{s['Speaker']} <NA> <NA>\n")

from google.colab import files
files.download(out_rttm_8b)
```
